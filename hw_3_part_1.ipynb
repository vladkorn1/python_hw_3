```python
import pandas as pd
```

```python
df = pd.read_csv('/Users/vladkornilov/Desktop/вшэ/Python_Module_1/hw_3_python/Electronic_sales_Sep2023-Sep2024.csv')
df.head(3)
```

```python
df2 = df[['Customer ID', 'Payment Method', 'Order Status', 'Total Price', 'Add-on Total']]
df2.head()
```

```python
df2["Order Status"].value_counts()
```

```python
completed_df2 = df2[df2['Order Status'] == 'Completed']
completed_df2.head()
```

```python
total_spent = completed_df2.groupby('Customer ID')['Total Price'].sum()
total_spent
```

```python
addons_spent = completed_df2.groupby('Customer ID')['Add-on Total'].sum()
addons_spent
```

```python
orders_count = completed_df2.groupby('Customer ID')['Order Status'].count()
orders_count
```

```python
preferred_payment = completed_df2.groupby(['Customer ID', 'Payment Method']).size()
preferred_payment
```

```python
preferred_payment = preferred_payment.reset_index()
preferred_payment
```

```python
preferred_payment = preferred_payment.rename(columns={0: 'count'})
preferred_payment
```

```python
preferred_payment = preferred_payment.sort_values(
    ['Customer ID', 'count', 'Payment Method'],
    ascending = [True, False, True]
)
preferred_payment
```

```python
preferred_payment = preferred_payment.drop_duplicates('Customer ID')
preferred_payment
```

```python
preferred_payment = preferred_payment[
    ["Customer ID", "Payment Method"]
]
preferred_payment
```

```python
preferred_payment = preferred_payment.rename(columns={'Payment Method': 'preferred_payment'})
preferred_payment
```

```python
total_spent = total_spent.reset_index()
total_spent = total_spent.rename(columns={'Total Price': 'total_spent'})
total_spent
```

```python
addons_spent = addons_spent.reset_index()
```

```python
addons_spent = addons_spent.rename(columns={'Add-on Total': 'addons_spent'})
```

```python
orders_count = orders_count.reset_index()
```

```python
orders_count = orders_count.rename(columns={'Order Status': 'orders_count'})
```

```python
final_df = preferred_payment.merge(
    total_spent, on = 'Customer ID'
)
```

```python
final_df = final_df.merge(
    addons_spent, on = 'Customer ID'
)
```

```python
final_df = final_df.merge(
    orders_count, on = 'Customer ID'
)
```

```python
final_df = final_df[['Customer ID', 'preferred_payment', 'total_spent', 'addons_spent', 'orders_count']]
```

```python
final_df = final_df.sort_values('total_spent', ascending = False)
```
